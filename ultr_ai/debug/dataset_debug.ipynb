```python
import pandas as pd
import numpy as np
import h5py
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import h5py
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, LabelEncoder, RobustScaler
from sklearn.impute import SimpleImputer, KNNImputer
import warnings
from typing import Dict, List, Tuple, Optional, Union
import logging
class MultimodalTBDataset(Dataset):
    def __init__(self, clinical_data_path: str, h5_file_path: str, site_data_path: str,
                 patient_ids: Optional[List[str]] = None, transform=None, 
                 fit_scalers: bool = True, clinical_scaler=None, is_training: bool = True,
                 imputation_strategy: str = 'median', use_robust_scaling: bool = False,
                 handle_outliers: bool = True, feature_selection: bool = False):
        
        self.clinical_data_path = clinical_data_path
        self.h5_file_path = h5_file_path
        self.site_data_path = site_data_path
        self.transform = transform
        self.is_training = is_training
        self.imputation_strategy = imputation_strategy
        self.use_robust_scaling = use_robust_scaling
        self.handle_outliers = handle_outliers
        self.feature_selection = feature_selection
        
        # Setup logging
        logging.basicConfig(level=logging.INFO)
        self.logger = logging.getLogger(__name__)
        
        # Load and process data
        self.clinical_data = self._load_clinical_data()
        self.site_data = self._load_site_data()
        self.patient_embeddings = self._load_patient_embeddings()
        
        # Determine patient IDs
        if patient_ids is not None:
            # Ensure patient IDs are strings
            self.patient_ids = [str(pid) for pid in patient_ids]
            self.logger.info(f"Using provided patient IDs: {len(self.patient_ids)}")
        else:
            self.patient_ids = self._find_common_patient_ids()
        
        # Process clinical features
        self.clinical_features, self.feature_schema, self.feature_order = self._process_clinical_features(
            fit_scalers, clinical_scaler
        )
        
        self.logger.info(f"Dataset initialized with {len(self.patient_ids)} patients")
        self.logger.info(f"Clinical features shape: {self.clinical_features.shape}")
        if self.patient_embeddings:
            sample_embedding = list(self.patient_embeddings.values())[0]
            self.logger.info(f"Patient embeddings dimension: {sample_embedding.shape}")
    
    def _load_clinical_data(self) -> pd.DataFrame:
        """Load and preprocess clinical data"""
        df = pd.read_csv(self.clinical_data_path)
        df['record_id'] = df['record_id'].astype(str)
        
        # Basic data validation
        self.logger.info(f"Loaded clinical data: {df.shape}")
        self.logger.info(f"Missing data percentage: {(df.isnull().sum().sum() / df.size) * 100:.2f}%")
        
        return df
    
    def _load_site_data(self) -> pd.DataFrame:
        """Load and aggregate site data"""
        df = pd.read_csv(self.site_data_path)
        df['patient_id'] = df['patient_id'].astype(str)
        
        # Enhanced aggregation with more robust statistics
        aggregation_dict = {
            'tb_label': 'first',
            'tb_logit': 'mean', 
            'tb_prob': 'mean',
            'tb_pred': 'first',
            'a_lines_finding': 'mean',
            'b_lines_finding': 'mean',
            'small_consolidations_finding': 'mean',
            'large_consolidations_finding': 'mean',
            'pleural_effusion_finding': 'mean',
            'mil_attention': 'mean'
        }
        
        # Filter to only include existing columns
        existing_agg_dict = {k: v for k, v in aggregation_dict.items() if k in df.columns}
        
        # Add count of observations per patient
        site_counts = df.groupby('patient_id').size().reset_index(name='site_observation_count')
        
        aggregated = df.groupby('patient_id').agg(existing_agg_dict).reset_index()
        aggregated = aggregated.merge(site_counts, on='patient_id', how='left')
        
        self.logger.info(f"Loaded site data: {len(df)} observations from {len(aggregated)} patients")
        
        return aggregated
    
    def _load_patient_embeddings(self) -> Dict[str, torch.Tensor]:
        """Load patient embeddings from H5 file with proper string conversion"""
        embeddings = {}
        try:
            with h5py.File(self.h5_file_path, 'r') as f:
                if 'patient_features' in f:
                    for pid in f['patient_features'].keys():
                        # Convert patient ID to string to ensure consistency
                        pid_str = str(pid)
                        embeddings[pid_str] = torch.tensor(
                            f['patient_features'][pid][:], dtype=torch.float32
                        )
                else:
                    self.logger.warning(f"'patient_features' not found in H5 file. Available keys: {list(f.keys())}")
                    
            self.logger.info(f"Loaded embeddings for {len(embeddings)} patients")
            
            # Debug: show sample patient IDs
            if embeddings:
                sample_ids = list(embeddings.keys())[:5]
                self.logger.info(f"Sample embedding patient IDs: {sample_ids}")
                
        except Exception as e:
            self.logger.error(f"Error loading embeddings: {e}")
            
        return embeddings
    
    def _find_common_patient_ids(self) -> List[str]:
        """Find patients with data in all modalities with proper debugging"""
        # Convert all IDs to strings for consistency
        clinical_ids = set(self.clinical_data['record_id'].astype(str))
        embedding_ids = set(self.patient_embeddings.keys())
        site_ids = set(self.site_data['patient_id'].astype(str))
        
        # Fixed the print statements
        print("Clinical ids", len(clinical_ids))
        print("Embedding ids", len(embedding_ids))
        print("Site ids", len(site_ids))
        
        # Find intersections
        common_ids = list(clinical_ids & embedding_ids & site_ids)
        
        # More detailed logging
        self.logger.info(f"Clinical IDs: {len(clinical_ids)}")
        self.logger.info(f"Embedding IDs: {len(embedding_ids)}")
        self.logger.info(f"Site IDs: {len(site_ids)}")
        self.logger.info(f"Common IDs: {len(common_ids)}")
        
        # Debug: show sample IDs for comparison
        if clinical_ids:
            sample_clinical = list(clinical_ids)[:3]
            self.logger.info(f"Sample clinical IDs: {sample_clinical}")
        
        if embedding_ids:
            sample_embedding = list(embedding_ids)[:3]
            self.logger.info(f"Sample embedding IDs: {sample_embedding}")
            
        if site_ids:
            sample_site = list(site_ids)[:3]
            self.logger.info(f"Sample site IDs: {sample_site}")
        
        # Show what's missing where
        clinical_only = clinical_ids - embedding_ids - site_ids
        embedding_only = embedding_ids - clinical_ids - site_ids
        site_only = site_ids - clinical_ids - embedding_ids
        
        if clinical_only:
            self.logger.warning(f"Clinical-only IDs (first 5): {list(clinical_only)[:5]}")
        if embedding_only:
            self.logger.warning(f"Embedding-only IDs (first 5): {list(embedding_only)[:5]}")
        if site_only:
            self.logger.warning(f"Site-only IDs (first 5): {list(site_only)[:5]}")
        
        return common_ids
    
    def _define_enhanced_feature_schema(self) -> Dict[str, List[str]]:
        """Enhanced feature schema based on CLUSSTER-Benin codebook"""
        return {
            'demographic': ['sexe', 'age'],
            'symptom': [
                'chief_complaint', 
                'fever_sympt_onset', 'cough_sympt_onset', 'dyspnea_sympt_onset', 
                'chestpain_symp_onset', 'hemoptysis_sympt_onset',
                'fever_acc_sympt_duration', 'cough_acc_sympt_duration', 'weight_acc_sympt_duration',
                'dyspnea_acc_sympt_duration', 'chest_acc_sympt_duration', 'hemopt_acc_sympt_duration',
                'sweats_acc_sympt_duration', 'asthen_acc_sympt_duration', 'anorex_acc_sympt_duration'
            ] + [f'accompanying_symptoms___{i}' for i in range(1, 14)] + [
                'chronic_cough', 'anthonissen_1', 'anthonissen_2', 'anthonissen_3'
            ],
            'vital_signs': [
                'poids_kg', 'taille_cm', 'temperature', 'frequence_cardiaque',
                'tension_arterielle_systol', 'tension_arterielle_diastol',
                'frequence_respiratoire', 'saturation_en_oxygene_spo2',
                'glasgow_coma_scale', 'quick_sofa_score', 'bmi'
            ],
            'medical_history': [
                'previous_tb_diagnosis', 'antibiotique_av', 'smoker', 'lung_diseases',
                'diabetes', 'hypertension', 'cardiopathy', 'hiv', 'covid_contact',
                'covid_vacc', 'traditional_med_sought'
            ],
            'physical_exam': [
                'general_state', 'pitting_peripheral_edema', 'cerv_and_or_axillary_adp',
                'auscultation_a'
            ],
            'laboratory': [
                'hiv_test_v2', 'cd4_counts', 'malaria_rdt', 'hb', 'mcv', 'thrombo',
                'leuko', 'neutro', 'lympho', 'tb_rif_genexpert', 'covid_pcr'
            ]
        }
    
    def _preprocess_clinical_features(self, df: pd.DataFrame) -> pd.DataFrame:
        """Advanced preprocessing of clinical features"""
        df = df.copy()
        
        # Gender encoding (1=Female, 2=Male -> 0=Female, 1=Male)
        if 'sexe' in df.columns:
            df['sexe'] = df['sexe'].map({1: 0, 2: 1})
        
        # Yes/No questions typically encoded as 1=Yes, 0=No
        binary_cols = ['chronic_cough', 'pitting_peripheral_edema', 'cerv_and_or_axillary_adp',
                      'lung_diseases', 'diabetes', 'hypertension', 'cardiopathy']
        for col in binary_cols:
            if col in df.columns:
                df[col] = df[col].map({1: 1, 0: 0}).fillna(0).astype(int)
        
        # HIV status (1=No, 2=Yes on treatment, 3=Yes not on treatment)
        if 'hiv' in df.columns:
            df['hiv'] = df['hiv'].map({1: 0, 2: 1, 3: 2})
        
        # Antibiotics (1=No, 2=Yes stopped, 3=Yes ongoing)
        if 'antibiotique_av' in df.columns:
            df['antibiotique_av'] = df['antibiotique_av'].map({1: 0, 2: 1, 3: 2})
        
        # General state (1-5 scale, higher is worse)
        if 'general_state' in df.columns:
            df['general_state'] = (df['general_state'] - 1) / 4
        
        return df
    
    def _handle_outliers(self, df: pd.DataFrame, method: str = 'iqr') -> pd.DataFrame:
        """Handle outliers in clinical data"""
        if not self.handle_outliers:
            return df
            
        df = df.copy()
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        
        for col in numeric_cols:
            if method == 'iqr':
                Q1 = df[col].quantile(0.25)
                Q3 = df[col].quantile(0.75)
                IQR = Q3 - Q1
                lower_bound = Q1 - 1.5 * IQR
                upper_bound = Q3 + 1.5 * IQR
                
                # Cap outliers instead of removing them
                df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)
            
            elif method == 'zscore':
                mean = df[col].mean()
                std = df[col].std()
                df[col] = df[col].clip(lower=mean - 3*std, upper=mean + 3*std)
        
        return df
    
    def _process_clinical_features(self, fit_scalers: bool = True, 
                                 clinical_scaler=None) -> Tuple[pd.DataFrame, Dict, List]:
        """
        FIXED: Enhanced clinical feature processing with proper error handling
        """
        schema = self._define_enhanced_feature_schema()
        all_features = []
        for domain_features in schema.values():
            all_features.extend(domain_features)
        
        # Filter available features - CRITICAL FIX: Only use features that actually exist
        available_features = [f for f in all_features if f in self.clinical_data.columns]
        missing_features = set(all_features) - set(available_features)
        
        if missing_features:
            self.logger.warning(f"Missing {len(missing_features)} expected features: {list(missing_features)[:10]}...")
        
        # CRITICAL FIX: Ensure we only select available features
        try:
            df = self.clinical_data[['record_id'] + available_features].copy()
            df = df.set_index('record_id')
            self.logger.info(f"Selected {len(available_features)} available features from {len(all_features)} expected")
        except KeyError as e:
            self.logger.error(f"Error selecting features: {e}")
            # Fallback: only select features that definitely exist
            safe_features = [f for f in available_features if f in self.clinical_data.columns]
            df = self.clinical_data[['record_id'] + safe_features].copy()
            df = df.set_index('record_id')
            available_features = safe_features
            self.logger.warning(f"Using fallback with {len(safe_features)} safe features")
        
        # Preprocess features
        df = self._preprocess_clinical_features(df)
        
        # Handle outliers
        df = self._handle_outliers(df)
        
        # Update schema to only include available features - CRITICAL FIX
        schema_filtered = {}
        for domain, features in schema.items():
            domain_available = [f for f in features if f in df.columns]
            if domain_available:  # Only include domains that have available features
                schema_filtered[domain] = domain_available
        
        # Create final feature order - CRITICAL FIX: Only use available features
        feature_order = []
        for domain in ['demographic', 'symptom', 'vital_signs', 'medical_history', 'physical_exam', 'laboratory']:
            if domain in schema_filtered:
                feature_order.extend(schema_filtered[domain])
        
        # CRITICAL FIX: Ensure feature_order only contains columns that exist in df
        feature_order = [f for f in feature_order if f in df.columns]
        
        # Reorder dataframe - CRITICAL FIX: Only select existing columns
        df = df[feature_order]
        
        self.logger.info(f"Final feature matrix shape before processing: {df.shape}")
        self.logger.info(f"Feature order length: {len(feature_order)}")
        
        # Advanced imputation
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        categorical_cols = df.select_dtypes(exclude=[np.number]).columns
        
        # Numeric imputation
        if len(numeric_cols) > 0:
            try:
                if self.imputation_strategy == 'knn':
                    imputer = KNNImputer(n_neighbors=5)
                    numeric_imputed = imputer.fit_transform(df[numeric_cols])
                    df[numeric_cols] = numeric_imputed
                else:
                    imputer = SimpleImputer(strategy=self.imputation_strategy)
                    numeric_imputed = imputer.fit_transform(df[numeric_cols])
                    df[numeric_cols] = numeric_imputed
                self.logger.info(f"Imputed {len(numeric_cols)} numeric features")
            except Exception as e:
                self.logger.error(f"Error in numeric imputation: {e}")
                # Fallback to simple mean imputation
                df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
        
        # Categorical imputation
        if len(categorical_cols) > 0:
            try:
                cat_imputer = SimpleImputer(strategy='most_frequent')
                categorical_imputed = cat_imputer.fit_transform(df[categorical_cols])
                df[categorical_cols] = categorical_imputed
                
                # Label encoding for categorical variables
                for col in categorical_cols:
                    le = LabelEncoder()
                    df[col] = le.fit_transform(df[col].astype(str))
                self.logger.info(f"Imputed and encoded {len(categorical_cols)} categorical features")
            except Exception as e:
                self.logger.error(f"Error in categorical imputation: {e}")
                # Fallback to simple mode imputation
                for col in categorical_cols:
                    df[col] = df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else 0)
                    le = LabelEncoder()
                    df[col] = le.fit_transform(df[col].astype(str))
        
        # CRITICAL FIX: Verify dimensions before scaling
        self.logger.info(f"DataFrame shape before scaling: {df.shape}")
        self.logger.info(f"Feature order length: {len(feature_order)}")
        
        # Ensure feature_order matches the actual DataFrame columns
        actual_columns = df.columns.tolist()
        if len(feature_order) != len(actual_columns):
            self.logger.warning(f"Feature order mismatch: expected {len(feature_order)}, got {len(actual_columns)}")
            feature_order = actual_columns
            self.logger.info(f"Using actual column order: {feature_order[:10]}...")
        
        # Scaling
        try:
            if fit_scalers:
                if self.use_robust_scaling:
                    scaler = RobustScaler()
                else:
                    scaler = StandardScaler()
                
                scaled_values = scaler.fit_transform(df)
                self.clinical_scaler = scaler
            else:
                if clinical_scaler is None:
                    raise ValueError("clinical_scaler must be provided if fit_scalers is False")
                scaler = clinical_scaler
                scaled_values = scaler.transform(df)
                self.clinical_scaler = scaler
            
            # CRITICAL FIX: Ensure dimensions match before creating DataFrame
            self.logger.info(f"Scaled values shape: {scaled_values.shape}")
            self.logger.info(f"Feature order length: {len(feature_order)}")
            
            if scaled_values.shape[1] != len(feature_order):
                self.logger.error(f"Dimension mismatch: scaled_values has {scaled_values.shape[1]} columns, but feature_order has {len(feature_order)} elements")
                # Use the actual number of columns from scaled_values
                feature_order = feature_order[:scaled_values.shape[1]]
                self.logger.warning(f"Truncated feature_order to match scaled_values: {len(feature_order)}")
            
            # Create final dataframe
            scaled_df = pd.DataFrame(scaled_values, index=df.index, columns=feature_order)
            
        except Exception as e:
            self.logger.error(f"Error in scaling: {e}")
            raise
        
        self.logger.info(f"Processed {len(feature_order)} clinical features")
        self.logger.info(f"Feature distribution by domain: {[(d, len(f)) for d, f in schema_filtered.items()]}")
        
        return scaled_df, schema_filtered, feature_order
    
    def get_feature_importance_scores(self) -> Dict[str, float]:
        """Calculate basic feature importance scores"""
        if not hasattr(self, 'clinical_features'):
            return {}
            
        importance_scores = {}
        
        for feature in self.feature_order:
            if feature in self.clinical_features.columns:
                variance_score = self.clinical_features[feature].var()
                completeness_score = 1 - (self.clinical_data[feature].isnull().sum() / len(self.clinical_data))
                importance_scores[feature] = 0.7 * variance_score + 0.3 * completeness_score
        
        return importance_scores
    
    def __len__(self) -> int:
        return len(self.patient_ids)
    
    def __getitem__(self, idx: int) -> Dict[str, torch.Tensor]:
        pid = self.patient_ids[idx]
        
        try:
            # Get site data
            site_rows = self.site_data[self.site_data['patient_id'] == pid]
            if len(site_rows) == 0:
                raise ValueError(f"No site data found for patient {pid}")
            site_row = site_rows.iloc[0]
            
            # Get clinical features
            if pid not in self.clinical_features.index:
                raise ValueError(f"No clinical features found for patient {pid}")
            clinical_vector = self.clinical_features.loc[pid].values
            
            # Get patient embedding
            if pid not in self.patient_embeddings:
                raise ValueError(f"No embedding found for patient {pid}")
            patient_embedding = self.patient_embeddings[pid]
            
            return {
                'patient_id': pid,
                'clinical_features': torch.tensor(clinical_vector, dtype=torch.float32),
                'patient_embedding': patient_embedding,
                'tb_label': torch.tensor(site_row.get('tb_label', 0.0), dtype=torch.float32),
                'tb_prob': torch.tensor(site_row.get('tb_prob', 0.0), dtype=torch.float32),
                'pathology_findings': torch.tensor([
                    site_row.get('a_lines_finding', 0.0), 
                    site_row.get('b_lines_finding', 0.0),
                    site_row.get('small_consolidations_finding', 0.0), 
                    site_row.get('large_consolidations_finding', 0.0),
                    site_row.get('pleural_effusion_finding', 0.0)
                ], dtype=torch.float32),
                'mil_attention': torch.tensor(site_row.get('mil_attention', 0.0), dtype=torch.float32),
                'feature_schema': self.feature_schema,
                'site_observation_count': torch.tensor(site_row.get('site_observation_count', 1), dtype=torch.float32)
            }
            
        except Exception as e:
            self.logger.error(f"Error loading data for patient {pid}: {e}")
            
            # Return dummy data to prevent dataset failure
            dummy_clinical = torch.zeros(len(self.feature_order), dtype=torch.float32)
            dummy_embedding = torch.zeros_like(list(self.patient_embeddings.values())[0]) if self.patient_embeddings else torch.zeros(512)
            
            return {
                'patient_id': pid,
                'clinical_features': dummy_clinical,
                'patient_embedding': dummy_embedding,
                'tb_label': torch.tensor(0.0, dtype=torch.float32),
                'tb_prob': torch.tensor(0.0, dtype=torch.float32),
                'pathology_findings': torch.zeros(5, dtype=torch.float32),
                'mil_attention': torch.tensor(0.0, dtype=torch.float32),
                'feature_schema': self.feature_schema,
                'site_observation_count': torch.tensor(1.0, dtype=torch.float32)
            }

def load_fold_split_robust(fold_csv_path: str) -> Tuple[List[str], List[str], List[str]]:
    """
    Robust fold splitting function that handles different CSV formats
    """
    df = pd.read_csv(fold_csv_path)
    
    # Format 1: split column with patient_id column
    if 'split' in df.columns and 'patient_id' in df.columns:
        print("📋 Using 'split' column format")
        train_ids = df[df["split"] == "train"]["patient_id"].astype(str).tolist()
        val_ids = df[df["split"] == "val"]["patient_id"].astype(str).tolist()
        test_ids = df[df["split"] == "test"]["patient_id"].astype(str).tolist()
        
    # Format 2: Separate columns for each split
    elif all(col in df.columns for col in ['train_ids', 'valid_ids', 'test_ids']):
        print("📋 Using separate ID columns format")
        train_ids = df['train_ids'].dropna().astype(str).tolist()
        val_ids = df['valid_ids'].dropna().astype(str).tolist()
        test_ids = df['test_ids'].dropna().astype(str).tolist()
        
    # Format 3: Different naming convention
    elif 'split' in df.columns:
        # Try to infer patient ID column
        id_cols = [col for col in df.columns if 'id' in col.lower() and col != 'split']
        if id_cols:
            patient_col = id_cols[0]
            print(f"📋 Using '{patient_col}' as patient ID column")
            train_ids = df[df["split"] == "train"][patient_col].astype(str).tolist()
            val_ids = df[df["split"] == "val"][patient_col].astype(str).tolist()
            test_ids = df[df["split"] == "test"][patient_col].astype(str).tolist()
        else:
            raise ValueError(f"Cannot find patient ID column in {df.columns}")
    
    else:
        raise ValueError(f"Unsupported CSV format. Columns: {list(df.columns)}")
    
    print(f"📊 Loaded splits - Train: {len(train_ids)}, Val: {len(val_ids)}, Test: {len(test_ids)}")
    
    return train_ids, val_ids, test_ids

import torch
from torch.utils.data import DataLoader
from typing import Dict, List, Any

def custom_collate_fn(batch: List[Dict[str, Any]]) -> Dict[str, Any]:
    """
    Custom collate function to properly handle feature_schema and other metadata
    """
    # Extract feature_schema from first item (they should all be the same)
    feature_schema = batch[0]['feature_schema']
    
    # Create the collated batch
    collated_batch = {
        'patient_id': [item['patient_id'] for item in batch],
        'clinical_features': torch.stack([item['clinical_features'] for item in batch]),
        'patient_embedding': torch.stack([item['patient_embedding'] for item in batch]),
        'tb_label': torch.stack([item['tb_label'] for item in batch]),
        'tb_prob': torch.stack([item['tb_prob'] for item in batch]),
        'pathology_findings': torch.stack([item['pathology_findings'] for item in batch]),
        'mil_attention': torch.stack([item['mil_attention'] for item in batch]),
        'site_observation_count': torch.stack([item['site_observation_count'] for item in batch]),
        'feature_schema': feature_schema  # Single copy, not replicated
    }
    
    return collated_batch


def create_dataloaders(clinical_data_path: str, base_h5_path: str, base_site_path: str,
                              fold_num: int, base_fold_path: str, batch_size: int = 32, 
                              num_workers: int = 4, **dataset_kwargs) -> Tuple:

    """
    Updated dataloader creation with custom collate function
    """
    paths = {
        'train': (f"{base_h5_path}/train_full_model_fold{fold_num}_complex_data.h5",
                  f"{base_site_path}/train_full_model_fold{fold_num}_sites.csv"),
        'val':   (f"{base_h5_path}/val_full_model_fold{fold_num}_complex_data.h5",
                  f"{base_site_path}/val_full_model_fold{fold_num}_sites.csv"),
        'test':  (f"{base_h5_path}/test_full_model_fold{fold_num}_complex_data.h5",
                  f"{base_site_path}/test_full_model_fold{fold_num}_sites.csv")
    }

    fold_csv_path = f'{base_fold_path}/Fold_{fold_num}.csv'
    train_ids, val_ids, test_ids = load_fold_split_robust(fold_csv_path)

    train_dataset = MultimodalTBDataset(clinical_data_path, *paths['train'], fit_scalers=True)
    val_dataset = MultimodalTBDataset(clinical_data_path, *paths['val'], fit_scalers=False,
                                      clinical_scaler=train_dataset.clinical_scaler)
    test_dataset = MultimodalTBDataset(clinical_data_path, *paths['test'], fit_scalers=False,
                                       clinical_scaler=train_dataset.clinical_scaler)

    # Use custom collate function
    return (
        DataLoader(train_dataset, batch_size=batch_size, shuffle=True, 
                  num_workers=num_workers, pin_memory=True, collate_fn=custom_collate_fn),
        DataLoader(val_dataset, batch_size=batch_size, shuffle=False, 
                  num_workers=num_workers, pin_memory=True, collate_fn=custom_collate_fn),
        DataLoader(test_dataset, batch_size=batch_size, shuffle=False, 
                  num_workers=num_workers, pin_memory=True, collate_fn=custom_collate_fn),
        train_dataset.clinical_scaler
    )

def analyze_improved_dataset(dataset: MultimodalTBDataset) -> Dict:
    """
    Analyze the improved dataset with enhanced metrics
    """
    # Basic dataset statistics
    tb_labels = []
    clinical_feature_stats = []
    pathology_stats = []
    
    for i in range(len(dataset)):
        sample = dataset[i]
        tb_labels.append(sample['tb_label'].item())
        clinical_feature_stats.append(sample['clinical_features'].numpy())
        pathology_stats.append(sample['pathology_findings'].numpy())
    
    tb_labels = np.array(tb_labels)
    clinical_features = np.array(clinical_feature_stats)
    pathology_features = np.array(pathology_stats)
    
    # Calculate detailed statistics
    analysis = {
        'dataset_size': len(dataset),
        'tb_positive_cases': int(np.sum(tb_labels == 1)),
        'tb_negative_cases': int(np.sum(tb_labels == 0)),
        'tb_positive_ratio': float(np.mean(tb_labels == 1)),
        'class_balance_ratio': float(np.sum(tb_labels == 1) / max(np.sum(tb_labels == 0), 1)),
        
        # Clinical features analysis
        'clinical_features': {
            'mean': clinical_features.mean(axis=0),
            'std': clinical_features.std(axis=0),
            'min': clinical_features.min(axis=0),
            'max': clinical_features.max(axis=0),
            'feature_names': dataset.feature_order
        },
        
        # Pathology findings analysis
        'pathology_findings': {
            'mean': pathology_features.mean(axis=0),
            'std': pathology_features.std(axis=0),
            'feature_names': ['a_lines', 'b_lines', 'small_consolidations', 
                            'large_consolidations', 'pleural_effusion']
        },
        
        # Feature importance
        'feature_importance': dataset.get_feature_importance_scores(),
        
        # Domain distribution
        'domain_distribution': {domain: len(features) 
                              for domain, features in dataset.feature_schema.items()}
    }
    
    # Print summary
    print(f"📊 Enhanced Dataset Analysis:")
    print(f"   Total patients: {analysis['dataset_size']}")
    print(f"   TB positive: {analysis['tb_positive_cases']} ({analysis['tb_positive_ratio']:.1%})")
    print(f"   TB negative: {analysis['tb_negative_cases']} ({1-analysis['tb_positive_ratio']:.1%})")
    print(f"   Class balance ratio: {analysis['class_balance_ratio']:.3f}")
    print(f"   Clinical features: {len(analysis['clinical_features']['feature_names'])}")
    print(f"   Domain distribution: {analysis['domain_distribution']}")
    
    return analysis
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
import warnings
from typing import Dict, List, Tuple, Optional
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

warnings.filterwarnings('ignore')

def comprehensive_data_analysis(clinical_data_path: str, site_data_path: str = None, 
                              save_plots: bool = True, plot_dir: str = "clinical_analysis_plots"):
    """
    Comprehensive analysis of clinical data with multiple visualizations
    """
    import os
    if save_plots:
        os.makedirs(plot_dir, exist_ok=True)
    
    # Load data
    df = pd.read_csv(clinical_data_path)
    if site_data_path:
        site_df = pd.read_csv(site_data_path)
        # Merge with site data if available
        df = df.merge(site_df, left_on='record_id', right_on='patient_id', how='left')
    
    print(f"🔍 Dataset Overview:")
    print(f"   Total patients: {len(df)}")
    print(f"   Features: {df.shape[1]}")
    print(f"   Missing data: {df.isnull().sum().sum()} total missing values")
    
    # 1. Demographics Analysis
    fig_demographics = make_subplots(
        rows=2, cols=2,
        subplot_titles=["Age Distribution", "Gender Distribution", "Age by Gender", "BMI Distribution"],
        specs=[[{"type": "histogram"}, {"type": "pie"}],
               [{"type": "box"}, {"type": "histogram"}]]
    )
    
    if 'age' in df.columns:
        # Age distribution
        fig_demographics.add_trace(
            go.Histogram(x=df['age'].dropna(), name="Age", nbinsx=20),
            row=1, col=1
        )
        
        # Age by gender box plot
        if 'sexe' in df.columns:
            gender_map = {1: 'Female', 2: 'Male'}
            df['gender_label'] = df['sexe'].map(gender_map)
            
            for gender in df['gender_label'].dropna().unique():
                fig_demographics.add_trace(
                    go.Box(y=df[df['gender_label'] == gender]['age'], name=gender),
                    row=2, col=1
                )
    
    if 'sexe' in df.columns:
        # Gender pie chart
        gender_counts = df['sexe'].value_counts()
        gender_labels = [gender_map.get(k, f'Unknown_{k}') for k in gender_counts.index]
        fig_demographics.add_trace(
            go.Pie(labels=gender_labels, values=gender_counts.values, name="Gender"),
            row=1, col=2
        )
    
    if 'bmi' in df.columns:
        # BMI distribution
        fig_demographics.add_trace(
            go.Histogram(x=df['bmi'].dropna(), name="BMI", nbinsx=25),
            row=2, col=2
        )
    
    fig_demographics.update_layout(height=800, title_text="📊 Demographics Analysis")
    if save_plots:
        fig_demographics.write_html(f"{plot_dir}/demographics_analysis.html")
    fig_demographics.show()
    
    # 2. Symptoms Analysis
    symptom_cols = [col for col in df.columns if 'sympt' in col.lower() or 'chief_complaint' in col]
    if symptom_cols:
        print(f"\n🩺 Symptoms Analysis ({len(symptom_cols)} symptom features)")
        
        # Symptom duration analysis
        duration_cols = [col for col in symptom_cols if 'duration' in col or 'onset' in col]
        if duration_cols:
            plt.figure(figsize=(15, 10))
            
            # Create subplots for symptom durations
            n_cols = 3
            n_rows = (len(duration_cols) + n_cols - 1) // n_cols
            
            for i, col in enumerate(duration_cols):
                plt.subplot(n_rows, n_cols, i + 1)
                data = df[col].dropna()
                if len(data) > 0:
                    plt.hist(data, bins=20, alpha=0.7, edgecolor='black')
                    plt.title(f"{col.replace('_', ' ').title()}")
                    plt.xlabel("Days")
                    plt.ylabel("Frequency")
            
            plt.tight_layout()
            plt.suptitle("Symptom Duration Distributions", y=1.02, fontsize=16)
            if save_plots:
                plt.savefig(f"{plot_dir}/symptom_durations.png", dpi=300, bbox_inches='tight')
            plt.show()
    
    # 3. Vital Signs Analysis
    vital_cols = ['temperature', 'frequence_cardiaque', 'tension_arterielle_systol', 
                  'tension_arterielle_diastol', 'frequence_respiratoire', 'saturation_en_oxygene_spo2']
    vital_cols = [col for col in vital_cols if col in df.columns]
    
    if vital_cols:
        print(f"\n💓 Vital Signs Analysis")
        
        # Correlation heatmap
        vital_data = df[vital_cols].select_dtypes(include=[np.number])
        if not vital_data.empty:
            plt.figure(figsize=(10, 8))
            correlation_matrix = vital_data.corr()
            sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
                       square=True, linewidths=0.5)
            plt.title("Vital Signs Correlation Matrix")
            plt.tight_layout()
            if save_plots:
                plt.savefig(f"{plot_dir}/vitals_correlation.png", dpi=300, bbox_inches='tight')
            plt.show()
            
            # Vital signs distributions
            fig_vitals = make_subplots(
                rows=2, cols=3,
                subplot_titles=[col.replace('_', ' ').title() for col in vital_cols[:6]]
            )
            
            for i, col in enumerate(vital_cols[:6]):
                row = (i // 3) + 1
                col_pos = (i % 3) + 1
                
                data = df[col].dropna()
                if len(data) > 0:
                    fig_vitals.add_trace(
                        go.Histogram(x=data, name=col, showlegend=False),
                        row=row, col=col_pos
                    )
            
            fig_vitals.update_layout(height=600, title_text="Vital Signs Distributions")
            if save_plots:
                fig_vitals.write_html(f"{plot_dir}/vitals_distributions.html")
            fig_vitals.show()
    
    # 4. Comorbidities Analysis
    comorbidity_cols = ['previous_tb_diagnosis', 'smoker', 'lung_diseases', 'diabetes', 
                       'hypertension', 'cardiopathy', 'hiv']
    comorbidity_cols = [col for col in comorbidity_cols if col in df.columns]
    
    if comorbidity_cols:
        print(f"\n🏥 Comorbidities Analysis")
        
        # Comorbidity prevalence
        comorbidity_prevalence = {}
        for col in comorbidity_cols:
            if df[col].dtype in ['object', 'bool'] or df[col].nunique() <= 10:
                # Categorical/binary variable
                value_counts = df[col].value_counts(normalize=True)
                # Assume positive values (1, 'Yes', etc.) indicate presence
                positive_values = [1, 'Yes', 'Oui', 'yes', 'oui', True, 2]  # 2 might be positive in some encodings
                prevalence = sum(value_counts.get(val, 0) for val in positive_values if val in value_counts.index)
                comorbidity_prevalence[col] = prevalence
        
        if comorbidity_prevalence:
            plt.figure(figsize=(12, 6))
            conditions = list(comorbidity_prevalence.keys())
            prevalences = list(comorbidity_prevalence.values())
            
            bars = plt.bar(range(len(conditions)), prevalences, color='skyblue', edgecolor='navy')
            plt.xlabel("Comorbidities")
            plt.ylabel("Prevalence")
            plt.title("Comorbidity Prevalence in Study Population")
            plt.xticks(range(len(conditions)), [c.replace('_', ' ').title() for c in conditions], 
                      rotation=45, ha='right')
            
            # Add prevalence percentages on bars
            for bar, prev in zip(bars, prevalences):
                plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
                        f'{prev:.1%}', ha='center', va='bottom')
            
            plt.tight_layout()
            if save_plots:
                plt.savefig(f"{plot_dir}/comorbidity_prevalence.png", dpi=300, bbox_inches='tight')
            plt.show()
    
    # 5. Outcome Analysis (if available)
    if 'tb_label' in df.columns:
        print(f"\n🎯 TB Outcome Analysis")
        
        tb_distribution = df['tb_label'].value_counts()
        
        # TB outcome distribution
        fig_outcome = go.Figure()
        fig_outcome.add_trace(go.Pie(
            labels=['TB Negative', 'TB Positive'], 
            values=[tb_distribution.get(0, 0), tb_distribution.get(1, 0)],
            hole=0.3,
            marker_colors=['lightgreen', 'lightcoral']
        ))
        fig_outcome.update_layout(title="TB Diagnosis Distribution")
        if save_plots:
            fig_outcome.write_html(f"{plot_dir}/tb_outcome_distribution.html")
        fig_outcome.show()
        
        # TB outcome by demographics
        if 'sexe' in df.columns and 'age' in df.columns:
            fig_demo_tb = make_subplots(
                rows=1, cols=2,
                subplot_titles=["TB by Gender", "TB by Age Group"],
                specs=[[{"type": "bar"}, {"type": "bar"}]]
            )
            
            # TB by gender
            tb_by_gender = df.groupby(['sexe', 'tb_label']).size().unstack(fill_value=0)
            if not tb_by_gender.empty:
                for tb_status in tb_by_gender.columns:
                    label = 'TB Positive' if tb_status == 1 else 'TB Negative'
                    fig_demo_tb.add_trace(
                        go.Bar(x=['Female', 'Male'], y=tb_by_gender[tb_status].values, 
                              name=label),
                        row=1, col=1
                    )
            
            # TB by age group
            df['age_group'] = pd.cut(df['age'], bins=[0, 30, 50, 70, 100], 
                                   labels=['18-30', '31-50', '51-70', '70+'])
            tb_by_age = df.groupby(['age_group', 'tb_label']).size().unstack(fill_value=0)
            if not tb_by_age.empty:
                for tb_status in tb_by_age.columns:
                    label = 'TB Positive' if tb_status == 1 else 'TB Negative'
                    fig_demo_tb.add_trace(
                        go.Bar(x=tb_by_age.index.astype(str), y=tb_by_age[tb_status].values, 
                              name=label, showlegend=False),
                        row=1, col=2
                    )
            
            fig_demo_tb.update_layout(height=400, title_text="TB Outcomes by Demographics")
            if save_plots:
                fig_demo_tb.write_html(f"{plot_dir}/tb_by_demographics.html")
            fig_demo_tb.show()
    
    # 6. Missing Data Analysis
    print(f"\n🔍 Missing Data Analysis")
    
    missing_data = df.isnull().sum()
    missing_percentage = (missing_data / len(df)) * 100
    missing_df = pd.DataFrame({
        'Missing_Count': missing_data,
        'Missing_Percentage': missing_percentage
    }).sort_values('Missing_Percentage', ascending=False)
    
    # Plot missing data for features with >5% missing
    significant_missing = missing_df[missing_df['Missing_Percentage'] > 5]
    
    if not significant_missing.empty:
        plt.figure(figsize=(12, 8))
        y_pos = np.arange(len(significant_missing))
        plt.barh(y_pos, significant_missing['Missing_Percentage'], color='coral')
        plt.yticks(y_pos, [col.replace('_', ' ').title() for col in significant_missing.index])
        plt.xlabel('Missing Percentage (%)')
        plt.title('Features with >5% Missing Data')
        plt.gca().invert_yaxis()
        
        # Add percentage labels
        for i, v in enumerate(significant_missing['Missing_Percentage']):
            plt.text(v + 0.5, i, f'{v:.1f}%', va='center')
        
        plt.tight_layout()
        if save_plots:
            plt.savefig(f"{plot_dir}/missing_data_analysis.png", dpi=300, bbox_inches='tight')
        plt.show()
    
    # 7. Feature Summary Statistics
    print(f"\n📈 Summary Statistics")
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    summary_stats = df[numeric_cols].describe()
    
    print("Key numeric features summary:")
    key_features = ['age', 'bmi', 'temperature', 'frequence_cardiaque', 'saturation_en_oxygene_spo2']
    available_key_features = [f for f in key_features if f in summary_stats.columns]
    
    if available_key_features:
        print(summary_stats[available_key_features].round(2))
    
    return {
        'summary_stats': summary_stats,
        'missing_data': missing_df,
        'comorbidity_prevalence': comorbidity_prevalence if 'comorbidity_prevalence' in locals() else {},
        'feature_schema': {
            'demographic': ['sexe', 'age'],
            'symptom': [col for col in df.columns if 'sympt' in col.lower() or 'chief_complaint' in col],
            'vital': [col for col in vital_cols if col in df.columns],
            'comorbidity': [col for col in comorbidity_cols if col in df.columns]
        }
    }


def validate_clinical_data(df: pd.DataFrame) -> Dict[str, List[str]]:
    """
    Validate clinical data and return issues found
    """
    issues = {
        'range_violations': [],
        'impossible_values': [],
        'missing_critical': [],
        'data_type_issues': []
    }
    
    # Define expected ranges for clinical variables
    clinical_ranges = {
        'age': (18, 120),
        'bmi': (10, 80),
        'temperature': (30, 45),
        'frequence_cardiaque': (30, 200),
        'tension_arterielle_systol': (60, 250),
        'tension_arterielle_diastol': (30, 150),
        'frequence_respiratoire': (5, 60),
        'saturation_en_oxygene_spo2': (50, 100),
        'glasgow_coma_scale': (3, 15),
        'poids_kg': (20, 300),
        'taille_cm': (100, 250)
    }
    
    # Check ranges
    for col, (min_val, max_val) in clinical_ranges.items():
        if col in df.columns:
            out_of_range = df[(df[col] < min_val) | (df[col] > max_val)][col].count()
            if out_of_range > 0:
                issues['range_violations'].append(f"{col}: {out_of_range} values out of range [{min_val}, {max_val}]")
    
    # Check for impossible combinations
    if 'tension_arterielle_systol' in df.columns and 'tension_arterielle_diastol' in df.columns:
        impossible_bp = df[df['tension_arterielle_systol'] <= df['tension_arterielle_diastol']]
        if len(impossible_bp) > 0:
            issues['impossible_values'].append(f"Blood pressure: {len(impossible_bp)} cases where systolic <= diastolic")
    
    # Check critical missing data
    critical_fields = ['age', 'sexe', 'record_id']
    for field in critical_fields:
        if field in df.columns:
            missing_count = df[field].isnull().sum()
            if missing_count > 0:
                issues['missing_critical'].append(f"{field}: {missing_count} missing values")
    
    return issues


class ClinicalEncoder(nn.Module):
    def __init__(self, clinical_dim: int, output_dim: int, feature_schema: Dict[str, List[str]], 
                 hidden_dim: int = 256, dropout: float = 0.1, domain_aware: bool = True,
                 use_attention: bool = True, use_batch_norm: bool = True):
        super().__init__()
        
        self.domain_aware = domain_aware
        self.use_attention = use_attention
        self.use_batch_norm = use_batch_norm
        self.feature_schema = feature_schema
        
        if self.domain_aware:
            self.feature_dims = {domain: len(features) for domain, features in feature_schema.items()}
            total_expected_dim = sum(self.feature_dims.values())
            assert clinical_dim == total_expected_dim, f"Mismatch: clinical_dim={clinical_dim}, expected={total_expected_dim}"
            
            # Domain-specific encoders with different architectures based on domain characteristics
            self.demographic_encoder = self._build_encoder(
                self.feature_dims['demographic'], hidden_dim // 4, dropout, layers=2
            )
            
            # Symptoms are complex and numerous, so use larger network
            self.symptom_encoder = self._build_encoder(
                self.feature_dims['symptom'], hidden_dim // 2, dropout, layers=3
            )
            
            # Vitals are numeric and well-structured
            self.vital_encoder = self._build_encoder(
                self.feature_dims['vital'], hidden_dim // 4, dropout, layers=2
            )
            
            # Other features (comorbidities, etc.)
            self.other_encoder = self._build_encoder(
                self.feature_dims['other'], hidden_dim // 4, dropout, layers=2
            )
            
            # Cross-domain attention mechanism
            if self.use_attention:
                self.domain_attention = nn.MultiheadAttention(
                    embed_dim=hidden_dim, num_heads=8, dropout=dropout, batch_first=True
                )
                self.attention_norm = nn.LayerNorm(hidden_dim)
            
            # Final fusion layer
            fusion_input_dim = hidden_dim * 2 if self.use_attention else hidden_dim
            self.fusion_layer = self._build_encoder(
                fusion_input_dim, output_dim, dropout, layers=3, final_activation=False
            )
            
        else:
            # Simple encoder for non-domain-aware processing
            self.encoder = self._build_encoder(
                clinical_dim, output_dim, dropout, layers=4, final_activation=False
            )

    def _build_encoder(self, input_dim: int, output_dim: int, dropout: float, 
                      layers: int = 3, final_activation: bool = True) -> nn.Module:
        """Build a flexible encoder with configurable layers"""
        modules = []
        
        current_dim = input_dim
        for i in range(layers - 1):
            next_dim = max(output_dim, current_dim // 2)
            modules.extend([
                nn.Linear(current_dim, next_dim),
                nn.BatchNorm1d(next_dim) if self.use_batch_norm else nn.Identity(),
                nn.ReLU(),
                nn.Dropout(dropout)
            ])
            current_dim = next_dim
        
        # Final layer
        modules.append(nn.Linear(current_dim, output_dim))
        if final_activation:
            if self.use_batch_norm:
                modules.append(nn.BatchNorm1d(output_dim))
            modules.extend([nn.ReLU(), nn.Dropout(dropout)])
        
        return nn.Sequential(*modules)
    
    def forward(self, clinical_features: torch.Tensor) -> torch.Tensor:
        if self.domain_aware:
            # Split features by domain
            domain_features = self._split_features_by_domain(clinical_features)
            
            # Encode each domain
            demo_encoded = self.demographic_encoder(domain_features['demographic'])
            symptom_encoded = self.symptom_encoder(domain_features['symptom'])
            vital_encoded = self.vital_encoder(domain_features['vital'])
            other_encoded = self.other_encoder(domain_features['other'])
            
            # Combine domain encodings
            combined = torch.cat([demo_encoded, symptom_encoded, vital_encoded, other_encoded], dim=1)
            
            # Apply cross-domain attention if enabled
            if self.use_attention:
                # Reshape for attention: (batch, seq_len=4, hidden_dim/4)
                batch_size = combined.size(0)
                domain_embeddings = torch.stack([
                    demo_encoded, symptom_encoded, vital_encoded, other_encoded
                ], dim=1)  # (batch, 4, hidden_dim/4)
                
                # Self-attention across domains
                attended, _ = self.domain_attention(
                    domain_embeddings, domain_embeddings, domain_embeddings
                )
                attended = self.attention_norm(attended + domain_embeddings)
                
                # Flatten and concatenate with original
                attended_flat = attended.flatten(1)
                combined = torch.cat([combined, attended_flat], dim=1)
            
            return self.fusion_layer(combined)
        else:
            return self.encoder(clinical_features)
    
    def _split_features_by_domain(self, clinical_features: torch.Tensor) -> Dict[str, torch.Tensor]:
        """Split clinical features into domain-specific tensors"""
        domain_features = {}
        start = 0
        
        for domain in ['demographic', 'symptom', 'vital', 'other']:
            end = start + self.feature_dims[domain]
            domain_features[domain] = clinical_features[:, start:end]
            start = end
            
        return domain_features
```
